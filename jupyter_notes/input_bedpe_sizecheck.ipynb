```python
import pandas as pd
```

```python
base_path = "/usr/users/papantonis1/aman/microc_project/loop_calling_premade_hic/merged_loops_3tools/"
```

```python
files = {
    "Peakachu": base_path + "peakachu_merged_loops_rbp1.bedpe",
    "Mustache": base_path + "mustache_merged_loops_rbp1_fixed_clean.bedpe",
    "Cooldots": base_path + "cooldots_merged_loops_rbp1.bedpe"
}
```

```python
for name, filepath in files.items():
    df = pd.read_csv(filepath, sep='\t', header=None, usecols=[1, 2, 4, 5])
    mid1 = (df[1] + df[2]) // 2
    mid2 = (df[4] + df[5]) // 2
    loop_sizes_kb = abs(mid2 - mid1) / 1000
    print(f"Summary for {name}:")
    print(loop_sizes_kb.describe())
    print("\n")
```
