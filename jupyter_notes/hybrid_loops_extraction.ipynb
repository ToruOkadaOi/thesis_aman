```python
import pandas as pd
from pybedtools import BedTool

# Load mega looplist
mega_loop = pd.read_csv(
    "/usr/users/papantonis1/aman/microc_project/loop_calling_premade_hic/merged_loops_3tools/megalooplist/mega_merged_loops_ctrl.bedpe",
    sep="\t", header=None,
    names=["chr1", "start1", "end1", "chr2", "start2", "end2"]
)

# Load peaks
ctcf_bed = BedTool("/usr/users/papantonis1/aman/microc_data/nadine_macro/C_CTCF_seacr_top0.01.peaks.stringent.bed")
h3k27me3_bed = BedTool("/usr/users/papantonis1/aman/microc_data/cut_n_tag/nadine_cut_tag/nadine_cut_tag/C_H3K27me3_results/peakCalling/C_H3K27me3_seacr_top0.01.peaks.stringent.bed")
```

```python
# Extract anchor 1 and anchor 2 with index
anchor1_df = mega_loop[["chr1", "start1", "end1"]].copy()
anchor1_df["idx"] = anchor1_df.index
anchor1 = BedTool.from_dataframe(anchor1_df)

anchor2_df = mega_loop[["chr2", "start2", "end2"]].copy()
anchor2_df["idx"] = anchor2_df.index
anchor2 = BedTool.from_dataframe(anchor2_df)
```

```python
# Intersect anchor1 with CTCF, anchor2 with H3K27me3
a1_ctcf = anchor1.intersect(ctcf_bed, wa=True)
a2_h3k27 = anchor2.intersect(h3k27me3_bed, wa=True)

# Intersect anchor1 with H3K27me3, anchor2 with CTCF
a1_h3k27 = anchor1.intersect(h3k27me3_bed, wa=True)
a2_ctcf = anchor2.intersect(ctcf_bed, wa=True)
```

```python
# Extract indices
a1_ctcf_idx = set(int(f.fields[3]) for f in a1_ctcf)
a2_h3k27_idx = set(int(f.fields[3]) for f in a2_h3k27)

a1_h3k27_idx = set(int(f.fields[3]) for f in a1_h3k27)
a2_ctcf_idx = set(int(f.fields[3]) for f in a2_ctcf)
```

```python
# Hybrid index: one anchor CTCF, other H3K27me3
hybrid_idx = sorted((a1_ctcf_idx & a2_h3k27_idx) | (a1_h3k27_idx & a2_ctcf_idx))

# Save output
mega_loop.iloc[hybrid_idx].to_csv(
    "hybrid_CTCF_H3K27me3_loops_ctrl.bedpe", sep="\t", index=False, header=False
)

len(hybrid_idx)
```
