```python
import pandas as pd
from pybedtools import BedTool
```

```python
# loading the merged loops file
mega_loop = pd.read_csv("/usr/users/papantonis1/aman/microc_project/loop_calling_premade_hic/merged_loops_3tools/megalooplist/mega_merged_loops_rbp1.bedpe", sep="\t", header=None, names=["chr1", "start1", "end1", "chr2", "start2", "end2"])
```

```python
ctcf_bed = BedTool("/usr/users/papantonis1/aman/microc_data/nadine_macro/CPI_CTCF_seacr_top0.01.peaks.stringent.bed")
```

```python
anchor1_df = mega_loop[["chr1", "start1", "end1"]].copy()
anchor1_df["idx"] = anchor1_df.index
anchor1 = BedTool.from_dataframe(anchor1_df)
```

```python
anchor2_df = mega_loop[["chr2", "start2", "end2"]].copy()
anchor2_df["idx"] = anchor2_df.index
anchor2 = BedTool.from_dataframe(anchor2_df)
```

```python
#anchor1.head()
```

```python
anchor1_ctcf = anchor1.intersect(ctcf_bed, wa=True)
anchor2_ctcf = anchor2.intersect(ctcf_bed, wa=True)
```

```python
print(anchor1_ctcf)
```

```python
anchor1_idx = set(int(f.fields[3]) for f in anchor1_ctcf)
anchor2_idx = set(int(f.fields[3]) for f in anchor2_ctcf)
```

```python
common_idx = sorted(anchor1_idx & anchor2_idx) # both anchor overlap
common_idx
```

```python
mega_loop.iloc[common_idx].to_csv("ctcf_associated_loops_bothanchoroverlap_rbp1.bedpe", sep="\t", index=False, header=False)
```

```python
len(mega_loop.iloc[common_idx])
```

```python
either_idx = sorted(anchor1_idx | anchor2_idx) # either or
mega_loop.iloc[either_idx].to_csv("ctcf_associated_loops_eitheranchoroverlap_rbp1.bedpe", sep="\t", index=False, header=False)
```
