```python
import pandas as pd
from pybedtools import BedTool
```

```python
# import all loops
all_loops = pd.read_csv("/usr/users/papantonis1/aman/microc_project/loop_calling_premade_hic/jupyter_notes/ml_3_multiclassmodel.csv")
all_loops = all_loops.drop(columns=["Unnamed: 0"])
all_loops
```

```python
# Save left anchor BED
all_loops[['chr1', 'start1', 'end1']].to_csv("ml_3_multiclassmodel_left_anchors.bed", sep='\t', header=False, index=False)

# Save right anchor BED
all_loops[['chr2', 'start2', 'end2']].to_csv("ml_3_multiclassmodel_right_anchors.bed", sep='\t', header=False, index=False)
```

```python
# refgen
genome_fasta = "/usr/users/papantonis1/aman/microc_project/refgen/hg38.fa"
```

```python
# Left
left_bt = BedTool("ml_3_multiclassmodel_left_anchors.bed")
left_gc = left_bt.nucleotide_content(fi=genome_fasta).to_dataframe()
left_gc
```

```python
# Right
right_bt = BedTool("ml_3_multiclassmodel_right_anchors.bed")
right_gc = right_bt.nucleotide_content(fi=genome_fasta).to_dataframe()
right_gc
```

```python
left_gc_clean = left_gc.iloc[1:].reset_index(drop=True)
right_gc_clean = right_gc.iloc[1:].reset_index(drop=True)
```

```python
all_loops["GC_left"] = left_gc_clean["score"]
all_loops["GC_right"] = right_gc_clean["score"]
all_loops
```

```python
# convert to numeric
all_loops["GC_left"] = all_loops["GC_left"].astype(float)
all_loops["GC_right"] = all_loops["GC_right"].astype(float)

all_loops["GC_mean"] = (all_loops["GC_left"] + all_loops["GC_right"]) / 2
all_loops["GC_diff"] = (all_loops["GC_left"] - all_loops["GC_right"]).abs()
all_loops
```

```python
all_loops = all_loops.drop(columns=["GC_left", "GC_right"])
all_loops
```

```python
all_loops.to_csv("/usr/users/papantonis1/aman/microc_project/loop_calling_premade_hic/jupyter_notes/ml_4_multiclassmodel.csv")
```
