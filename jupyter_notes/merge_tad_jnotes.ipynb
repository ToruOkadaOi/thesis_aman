```python
import pandas as pd

ctrl_df = pd.read_csv("intra_inter_tad_ctrl.csv")
rbp1_df = pd.read_csv("intra_inter_tad_rbp1.csv")
```

```python
key_cols = ["chr1", "start1", "end1", "chr2", "start2", "end2"]

# Keep only the columns you need
ctrl_features = ctrl_df[key_cols + ["loop_class"]].copy()
rbp1_features = rbp1_df[key_cols + ["loop_class"]].copy()

# Add a source tag to distinguish them
ctrl_features["source"] = "ctrl"
rbp1_features["source"] = "rbp1"

# Concatenate the two DataFrames row-wise
merged_df = pd.concat([ctrl_features, rbp1_features], ignore_index=True)

merged_df
```

```python
merged_ml = pd.read_csv("/usr/users/papantonis1/aman/microc_project/loop_calling_premade_hic/jupyter_notes/merged_loops.csv")
merged_ml
```

```python
# merge
merged_df_dedup = merged_df.drop_duplicates(subset=key_cols)
merged_tad = merged_ml.merge(merged_df_dedup, on=key_cols, how="left")
```

```python
assert len(merged_tad) == len(merged_ml)
merged_tad["loop_class"].value_counts(dropna=False)
```
