```python
import subprocess

def run_coolpuppy(
    mcool_path,
    bedpe_file,
    output_file,
    mindist=None,
    maxdist=None,
    seed=None,
    threads=1
):
    """
    Run coolpup.py using the 'cool_notebook' conda environment via subprocess.
    """
    coolpup_script = "/usr/users/papantonis1/anaconda3/envs/cool_notebook/bin/coolpup.py"

    # Build optional arguments
    opts = []
    if mindist is not None:
        opts += [f"--mindist {mindist}"]
    if maxdist is not None:
        opts += [f"--maxdist {maxdist}"]
    if seed is not None:
        opts += [f"--seed {seed}"]
    if threads > 1:
        opts += [f"-p {threads}"]

    # Compose full command
    cmd = [
        "bash", "-c",
        f"source ~/anaconda3/etc/profile.d/conda.sh && conda activate cool_notebook && "
        f"python {coolpup_script} --format bedpe {mcool_path} {bedpe_file} "
        f"{' '.join(opts)} -o {output_file}"
    ]

    subprocess.run(cmd, check=True)
```

```python
mcool_path = "/usr/users/papantonis1/aman/microc_data/nadine_macro/ctrl-MNase-R1-filtered.mcool::resolutions/1000000"
```

```python
run_coolpuppy(
    mcool_path,
    "/usr/users/papantonis1/fithic-master/fithic/utils/fithic_outputs/ctrl.trans.q01.final.bedpe",
    "ctrl_transloops_fithic.npz",
    threads=8
)
```

```python
import subprocess

def plot_pileup(input_npz, output_pdf):
    """
    Run plotpup.py using the 'cool_notebook' conda environment.
    """
    plotpup_script = "/usr/users/papantonis1/anaconda3/envs/cool_notebook/bin/plotpup.py"
    cmd = [
        "bash", "-c",
        f"source ~/anaconda3/etc/profile.d/conda.sh && conda activate cool_notebook && python {plotpup_script} --input_pups {input_npz} --output {output_pdf}"
    ]
    subprocess.run(cmd, check=True)
```
