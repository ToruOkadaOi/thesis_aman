```python
# doing it individually 
# H3K27me3 ctrl bigwig - gene upregulated
## using subprocess
import subprocess

result = subprocess.run([
    "computeMatrix", "scale-regions",
    "-S", "/usr/users/papantonis1/aman/microc_data/cut_n_tag/nadine_cut_tag/nadine_cut_tag/C_H3K27me3_results/Aligned_files/Bigwig_scaled/C_H3K27me3.bw_RPGC.bw",
    "-R", "/usr/users/papantonis1/aman/rnaseq_data/up_genes_gencodegtf.bed",
    "-b", "2000",       # kb upstream of TSS
    "-a", "2000",       # kb downstream of TES
    "-m", "2500",      # gene body length to scale to
    "--skipZeros",
    "--missingDataAsZero",
    "--outFileName", "genelvl_ctrl_H3K27me3_upregulated_geneBody_matrix.gz"
], capture_output=True, text=True)

print(result.stdout)
print("STDERR:", result.stderr)
if result.returncode != 0:
    print("Error: computeMatrix failed")

subprocess.run([
    "plotHeatmap",
    "-m", "genelvl_ctrl_H3K27me3_upregulated_geneBody_matrix.gz",
    "-out", "/usr/users/papantonis1/aman/microc_project/loop_calling_premade_hic/jupyter_notes/heatmap_deeptoools/genelvl_ctrl_H3K27me3_upregulated_gene_heatmap_gtfgencode.png",
    "--colorMap", "YlGnBu",
    "--heatmapHeight", "15",
    "--regionsLabel", "Upregulated genes",
    "--samplesLabel", "H3K27me3 signal",
    "--plotTitle", "H3K27me3 signal(control) at upregulated genes",
    "--heatmapWidth", "6",
    "--heatmapHeight", "30"
])
```
