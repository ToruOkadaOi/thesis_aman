```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/usr/users/papantonis1/aman/microc_project/loop_calling_premade_hic/dfd_dvsd_mult_res_adjusted_classified.txt", sep="\t")

counts = df['type'].value_counts()
ax = counts.plot(kind='bar', color='skyblue', edgecolor='black')

# Add counts on top of bars
for i, count in enumerate(counts):
    ax.text(i, count + 100, str(count), ha='center', va='bottom', fontsize=10)

# Remove spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.ylabel("Number of TADs")
plt.title("TAD Reorganization Types")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

pivot = df.pivot_table(index='type', columns='subtype', aggfunc='size', fill_value=0)
sns.heatmap(pivot, annot=True, fmt="d", cmap="YlGnBu")
plt.title("Subtype distribution")
plt.ylabel("Type")
plt.xlabel("Subtype")
plt.show()
```
