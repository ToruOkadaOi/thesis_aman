```python
import pandas as pd
```

```python
# gtf - genocde
gtf = pd.read_csv(
    "/usr/users/papantonis1/aman/rnaseq_data/gencode.v48.chr_patch_hapl_scaff.annotation.gtf",
    sep="\t", comment="#", header=None,
    names=["chr","source","feature","start","end","score","strand","frame","attribute"]
)
```

```python
gtf_genes = gtf[gtf["feature"]=="gene"]
gtf_genes
```

```python
gtf_genes["gene_id"] = gtf_genes["attribute"].str.extract(r'gene_id "([^"]+)"')
gtf_genes["gene_id"] = gtf_genes["gene_id"].str.split(".").str[0]
gtf_genes
```

```python
deg_ids = pd.read_csv("/usr/users/papantonis1/aman/rnaseq_data/upregulated_genes_deseq.txt", header=None, sep="\t")[0].tolist()
deg_ids
```

```python
deg_coords = gtf_genes[gtf_genes["gene_id"].isin(deg_ids)]
print(deg_coords.head())
```

```python
deg_coords
```

```python
bed = deg_coords.copy()
bed["gene_name"] = bed["attribute"].str.extract(r'gene_name "([^"]+)"')

# BED6: chrom, start(0-based), end, name, score, strand
bed6 = pd.DataFrame({
    "chrom": bed["chr"],
    "start": bed["start"] - 1,       # convert to 0-based
    "end":   bed["end"],
    "name":  bed["gene_name"].fillna(bed["gene_id"]).fillna("."),
    "score": 0,
    "strand": bed["strand"]
}).sort_values(["chrom","start","end"])

bed6.to_csv("up_genes_gencodegtf.bed", sep="\t", header=False, index=False)
```

```python
# downreg.
deg_ids = pd.read_csv("/usr/users/papantonis1/aman/rnaseq_data/downregulated_genes_deseq.txt", header=None, sep="\t")[0].tolist()
#deg_ids
deg_coords = gtf_genes[gtf_genes["gene_id"].isin(deg_ids)]
#print(deg_coords.head())
bed = deg_coords.copy()
bed["gene_name"] = bed["attribute"].str.extract(r'gene_name "([^"]+)"')
# BED6: chrom, start(0-based), end, name, score, strand
bed6 = pd.DataFrame({
    "chrom": bed["chr"],
    "start": bed["start"] - 1,       # convert to 0-based
    "end":   bed["end"],
    "name":  bed["gene_name"].fillna(bed["gene_id"]).fillna("."),
    "score": 0,
    "strand": bed["strand"]
}).sort_values(["chrom","start","end"])
bed6.to_csv("down_genes_gencodegtf.bed", sep="\t", header=False, index=False)
bed6
```

```python
# Keep only gene rows
genes = gtf[gtf["feature"]=="gene"].copy()

# Extract gene_type from attributes
genes["gene_type"] = genes["attribute"].str.extract(r'gene_type "([^"]+)"')

# Count unique types
counts = genes["gene_type"].value_counts()

print("Number of unique gene_types:", counts.nunique())
print(counts)
```

```python
import re

# Build a tidy annotation table from the GTF once
ann = gtf_genes[["chr","start","end","strand","attribute"]].copy()
ann["gene_id"]   = ann["attribute"].str.extract(r'gene_id "([^"]+)"')
ann["gene_id"]   = ann["gene_id"].str.replace(r"\.\d+$","",regex=True)  # strip version
ann["gene_name"] = ann["attribute"].str.extract(r'gene_name "([^"]+)"')
ann["gene_type"] = ann["attribute"].str.extract(r'gene_type "([^"]+)"')
ann = ann.drop(columns="attribute").dropna(subset=["gene_id"]).drop_duplicates("gene_id")
```

```python
# --- UP ---
up_ids = pd.read_csv("/usr/users/papantonis1/aman/rnaseq_data/upregulated_genes_deseq.txt",
                     header=None, sep="\t")[0].astype(str)
up_ids = up_ids.str.replace(r"\.\d+$","",regex=True)  # strip versions if present

up = ann[ann["gene_id"].isin(up_ids)].copy()

# BED6 (0-based start)
up_bed6 = pd.DataFrame({
    "chrom": up["chr"],
    "start": up["start"] - 1,
    "end":   up["end"],
    "name":  up["gene_name"].fillna(up["gene_id"]),
    "score": 0,
    "strand": up["strand"]
}).sort_values(["chrom","start","end"])
#up_bed6.to_csv("up_genes_gencodegtf.bed", sep="\t", header=False, index=False)
up.head()
```

```python
# --- DOWN ---
down_ids = pd.read_csv("/usr/users/papantonis1/aman/rnaseq_data/downregulated_genes_deseq.txt",
                       header=None, sep="\t")[0].astype(str)
down_ids = down_ids.str.replace(r"\.\d+$","",regex=True)

down = ann[ann["gene_id"].isin(down_ids)].copy()

down_bed6 = pd.DataFrame({
    "chrom": down["chr"],
    "start": down["start"] - 1,
    "end":   down["end"],
    "name":  down["gene_name"].fillna(down["gene_id"]),
    "score": 0,
    "strand": down["strand"]
}).sort_values(["chrom","start","end"])
down_bed6.to_csv("down_genes_gencodegtf.bed", sep="\t", header=False, index=False)
down.head()
```

```python
# Categorize by gene_type
up_cat   = up[["gene_id","gene_name","gene_type","chr","start","end","strand"]].copy()
down_cat = down[["gene_id","gene_name","gene_type","chr","start","end","strand"]].copy()

# Save annotated tables
up_cat.to_csv("upregulated_genes_annotated.tsv", sep="\t", index=False)
down_cat.to_csv("downregulated_genes_annotated.tsv", sep="\t", index=False)

# Summary counts
up_counts   = up_cat["gene_type"].value_counts().rename_axis("gene_type").reset_index(name="up_count")
down_counts = down_cat["gene_type"].value_counts().rename_axis("gene_type").reset_index(name="down_count")

summary = pd.merge(up_counts, down_counts, on="gene_type", how="outer").fillna(0).sort_values(
    by=["up_count","down_count"], ascending=False
)
#summary.to_csv("deg_biotype_summary.tsv", sep="\t", index=False)
summary
```

```python
# Get only lncRNAs
lnc_up   = up_cat[up_cat["gene_type"]=="lncRNA"]
lnc_down = down_cat[down_cat["gene_type"]=="lncRNA"]

print("Up lncRNAs:", len(lnc_up))
print("Down lncRNAs:", len(lnc_down))

# Save them
#lnc_up.to_csv("upregulated_lncRNAs.tsv", sep="\t", index=False)
#lnc_down.to_csv("downregulated_lncRNAs.tsv", sep="\t", index=False)

# Peek at top entries
lnc_up.head()
```
