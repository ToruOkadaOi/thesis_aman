```python
# fill Nan rows in gene_id col.
gene_loop["gene_id"]=gene_loop["gene_id"].fillna('').astype(str)
```

```python
# convert into a list ## The gene id col. is now a list
gene_loop["gene_list"]=gene_loop["gene_id"].str.split(',')
gene_loop
```

```python
# creating a new row for each gene in the df
edges = gene_loop[['loop_id', 'gene_list']].explode('gene_list')
```

```python
edges = edges[edges['gene_list'] != '']  # Remove blanks
```

```python
# col structuring for cytoscape
edges.rename(columns={'loop_id': 'source', 'gene_list': 'target'}, inplace=True)
edges['interaction'] = 'loop-gene'
edges
```

```python
from mygene import MyGeneInfo
mg = MyGeneInfo()

# Extract unique transcript IDs (remove version suffixes)
transcript_ids = edges['target'].str.replace(r'\.\d+$', '', regex=True).unique().tolist()

# Query mygene.info
gene_info = mg.querymany(transcript_ids, scopes='ensembl.transcript', fields='symbol', species='human')

# Build mapping
mapping = {entry['query']: entry.get('symbol', '') for entry in gene_info}

# Apply mapping (strip .X from ENSTs before mapping)
edges['hgnc_symbol'] = edges['target'].str.replace(r'\.\d+$', '', regex=True).map(mapping)
```

```python
edges['target'] = edges['hgnc_symbol']
edges.drop(columns=['hgnc_symbol'], inplace=True)
edges
```

```python
edges.drop_duplicates(subset=['source', 'target'], inplace=True)
edges['target'].replace('', pd.NA, inplace=True)
edges['interaction'].replace('', pd.NA, inplace=True)
edges.dropna(inplace=True)
edges
```

```python
# checking
print("Unique loops (source nodes):", edges['source'].nunique())
```

```python
edges.to_csv('cytoscape_input.csv', sep=',', index=False)
```
