```r
suppressPackageStartupMessages({
  library(AnnotationDbi); library(org.Hs.eg.db)
  library(GenomicFeatures); library(TxDb.Hsapiens.UCSC.hg38.knownGene)
  library(dplyr)
})

in_bed  <- "/usr/users/papantonis1/aman/rnaseq_data/downregulated_gene_level_genebodies_auxincpi.bed"
out_bed <- "/usr/users/papantonis1/aman/rnaseq_data/downregulated_genelevel_genebody_hg38_ucsc.bed"

ids <- read.delim(in_bed, header = FALSE, sep = "\t", stringsAsFactors = FALSE)[[4]]
ent <- ids  # already Entrez

txdb <- TxDb.Hsapiens.UCSC.hg38.knownGene
g <- genes(txdb)
g <- g[names(g) %in% ent]

sym <- mapIds(org.Hs.eg.db, keys = names(g), column = "SYMBOL",
              keytype = "ENTREZID", multiVals = "first")
mcols(g)$name <- ifelse(is.na(sym), names(g), unname(sym))

df <- as.data.frame(g)[, c("seqnames","start","end","strand","name")]
df <- df %>% group_by(name, seqnames) %>%
  summarise(start = min(start), end = max(end), strand = dplyr::first(strand), .groups="drop")

bed <- df %>% transmute(chrom=as.character(seqnames),
                        start0=as.integer(start-1L), end=as.integer(end),
                        name=name, score=0L, strand=as.character(strand)) %>%
        arrange(chrom,start0,end)

write.table(bed, out_bed, sep="\t", quote=FALSE, row.names=FALSE, col.names=FALSE)
```
