##### Upregulated

```python
import pandas as pd

# Load the mega upregulated list (common upregulated genes)
mega_up = pd.read_csv("/usr/users/papantonis1/aman/rnaseq_data/mega_upregulated.tsv", sep="\t")
common_genes_up = mega_up.iloc[:, 0].astype(str).str.strip()
# Load log2FC
exon_df = pd.read_csv("/usr/users/papantonis1/aman/rnaseq_data/upregulated_genes_exons_irnaseqCM.txt", sep="\t")
intron_df = pd.read_csv("/usr/users/papantonis1/aman/rnaseq_data/upregulated_genes_introns_irnaseqCM.txt", sep="\t")
deseq_df = pd.read_csv("/usr/users/papantonis1/aman/rnaseq_data/upregulated_genes_deseq.txt", sep="\t")

# Standardize column names
exon_df = exon_df.rename(columns={'log2FC': 'log2fc_exon'})
intron_df = intron_df.rename(columns={'log2FC': 'log2fc_intron'})
deseq_df = deseq_df.rename(columns={'log2FC': 'log2fc_deseq'})

exon_df = exon_df.rename(columns={'ensembl': 'hgnc_symbol'})
intron_df = intron_df.rename(columns={'ensembl': 'hgnc_symbol'})

# Merge based on common genes
merged_up = pd.DataFrame({'hgnc_symbol': common_genes_up})
merged_up = merged_up.merge(exon_df[['hgnc_symbol', 'log2fc_exon']], on='hgnc_symbol', how='left')
merged_up = merged_up.merge(intron_df[['hgnc_symbol', 'log2fc_intron']], on='hgnc_symbol', how='left')
merged_up = merged_up.merge(deseq_df[['hgnc_symbol', 'log2fc_deseq']], on='hgnc_symbol', how='left')
```

##### Downregulated

```python
# Load the mega upregulated list (common upregulated genes)
mega_down = pd.read_csv("/usr/users/papantonis1/aman/rnaseq_data/mega_downregulated.tsv", sep="\t")
common_genes_down = mega_down.iloc[:, 0].astype(str).str.strip()
# Load log2FC
exon_df = pd.read_csv("/usr/users/papantonis1/aman/rnaseq_data/downregulated_genes_exons_irnaseqCM.txt", sep="\t")
intron_df = pd.read_csv("/usr/users/papantonis1/aman/rnaseq_data/downregulated_genes_introns_irnaseqCM.txt", sep="\t")
deseq_df = pd.read_csv("/usr/users/papantonis1/aman/rnaseq_data/downregulated_genes_deseq.txt", sep="\t")

# Standardize column names
exon_df = exon_df.rename(columns={'log2FC': 'log2fc_exon'})
intron_df = intron_df.rename(columns={'log2FC': 'log2fc_intron'})
deseq_df = deseq_df.rename(columns={'log2FC': 'log2fc_deseq'})

exon_df = exon_df.rename(columns={'ensembl': 'hgnc_symbol'})
intron_df = intron_df.rename(columns={'ensembl': 'hgnc_symbol'})

# Merge based on common genes
merged_down = pd.DataFrame({'hgnc_symbol': common_genes_down})
merged_down = merged_down.merge(exon_df[['hgnc_symbol', 'log2fc_exon']], on='hgnc_symbol', how='left')
merged_down = merged_down.merge(intron_df[['hgnc_symbol', 'log2fc_intron']], on='hgnc_symbol', how='left')
merged_down = merged_down.merge(deseq_df[['hgnc_symbol', 'log2fc_deseq']], on='hgnc_symbol', how='left')
```

```python
merged_down
```

```python
valid_pairs = merged_down[['log2fc_exon', 'log2fc_intron']].dropna()
print("Valid exon–intron pairs:", len(valid_pairs))

valid_pairs = merged_down[['log2fc_exon', 'log2fc_deseq']].dropna()
print("Valid exon–deseq pairs:", len(valid_pairs))

valid_pairs = merged_down[['log2fc_intron', 'log2fc_deseq']].dropna()
print("Valid intron–deseq pairs:", len(valid_pairs))
```

```python
from scipy.stats import spearmanr

subset = merged_down[['log2fc_exon', 'log2fc_deseq']].dropna()
rho, p = spearmanr(subset['log2fc_exon'], subset['log2fc_deseq'])
print("Spearman (exon vs deseq):", rho, p)
```

```python
from scipy.stats import ttest_rel

subset = merged_down[['log2fc_intron', 'log2fc_deseq']].dropna()
t_stat, p = ttest_rel(subset['log2fc_intron'], subset['log2fc_deseq'])
print("Paired t-test (intron vs deseq):", t_stat, p)
```

```python
from scipy.stats import wilcoxon

subset = merged_down[['log2fc_intron', 'log2fc_deseq']].dropna()
w_stat, p = wilcoxon(subset['log2fc_intron'], subset['log2fc_deseq'])
print("Wilcoxon test (intron vs deseq):", w_stat, p)
```

```python
### intron vs exon

from scipy.stats import spearmanr

subset = merged_down[['log2fc_intron', 'log2fc_exon']].dropna()
rho, p = spearmanr(subset['log2fc_intron'], subset['log2fc_exon'])
print("Spearman (intron vs exon):", rho, p)
```

```python
from scipy.stats import ttest_rel

subset = merged_down[['log2fc_intron', 'log2fc_exon']].dropna()
t_stat, p = ttest_rel(subset['log2fc_intron'], subset['log2fc_exon'])
print("Paired t-test (intron vs exon):", t_stat, p)
```

```python
from scipy.stats import wilcoxon

subset = merged_down[['log2fc_intron', 'log2fc_exon']].dropna()
w_stat, p = wilcoxon(subset['log2fc_intron'], subset['log2fc_exon'])
print("Wilcoxon (intron vs exon):", w_stat, p)
```
